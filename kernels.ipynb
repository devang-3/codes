```python
import imageio.v2 as imageio
import numpy as np
from IPython.display import display
from PIL import Image
from convolution import apply_filter_to_image

kernels = {
    "identity": np.array([
        [0, 0, 0],
        [0, 1, 0],
        [0, 0, 0]
    ]),
    "laplacian_4": np.array([
        [0, -1, 0],
        [-1, 4, -1],
        [0, -1, 0]
    ]),
    "laplacian_8": np.array([
        [-1, -1, -1],
        [-1, 8, -1],
        [-1, -1, -1]
    ]),
    "sharpen": np.array([
        [0, -1, 0],
        [-1, 5, -1],
        [0, -1, 0]
    ]),
    "box_blur": np.ones((3, 3)) / 9,
    "gaussian_blur": np.array([
        [1, 2, 1],
        [2, 4, 2],
        [1, 2, 1]
    ]) / 16,
    "edge_horizontal": np.array([
        [-1, -1, -1],
        [0, 0, 0],
        [1, 1, 1]
    ]),
    "edge_vertical": np.array([
        [-1, 0, 1],
        [-1, 0, 1],
        [-1, 0, 1]
    ]),
    "sobel_horizontal": np.array([
        [-1, -2, -1],
        [0, 0, 0],
        [1, 2, 1]
    ]),
    "sobel_vertical": np.array([
        [-1, 0, 1],
        [-2, 0, 2],
        [-1, 0, 1]
    ]),
    "prewitt_horizontal": np.array([
        [-1, -1, -1],
        [0, 0, 0],
        [1, 1, 1]
    ]),
    "prewitt_vertical": np.array([
        [-1, 0, 1],
        [-1, 0, 1],
        [-1, 0, 1]
    ]),
    "gabor": np.array([
        [0, 0, 0, 0, 0],
        [0, 1, 2, 1, 0],
        [0, 2, 4, 2, 0],
        [0, 1, 2, 1, 0],
        [0, 0, 0, 0, 0]
    ]) / 16,
    "scharr_vertical": np.array([
        [-3, 0, 3],
        [-10, 0, 10],
        [-3, 0, 3]
    ]),
    "scharr_horizontal": np.array([
        [-3, -10, -3],
        [0, 0, 0],
        [3, 10, 3]
    ])
}

image = imageio.imread('D:\\Data Science\\codes\\deepmind.jpg')
# display(Image.fromarray(image))
```

```python
filtered_image = apply_filter_to_image(image, kernels["gaussian_blur"])
display(Image.fromarray(filtered_image))

```

```python
filtered_images = {}

for name, kernel in kernels.items():
    filtered_images[name] = apply_filter_to_image(image, kernel)
    print(name)
    display(Image.fromarray(filtered_images[name]))
```
